```python
import numpy as np
import pandas as pd
from sklearn.model_selection import GroupKFold
from sklearn.metrics import root_mean_squared_error
import lightgbm as lgb

# Optional: CatBoost improves categorical interaction accuracy significantly
try:
    from catboost import CatBoostRegressor
    HAS_CATBOOST = True
except ImportError:
    HAS_CATBOOST = False

train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
```

```python
# 1. Unified schedule calendar ONLY for the 8 target trains (fixes train/test shift!)
target_trains = sorted(test['train_n'].unique())
all_sched = pd.concat([
    train[train['train_n'].isin(target_trains)][['trip_date', 'train_n']],
    test[['trip_date', 'train_n']]
]).drop_duplicates()

cal = pd.DataFrame({'trip_date': sorted(all_sched['trip_date'].unique())})
cal['dt'] = pd.to_datetime(cal['trip_date'])
cal['dow'] = cal['dt'].dt.dayofweek
cal['month'] = cal['dt'].dt.month

s709 = set(all_sched[all_sched['train_n'] == 709]['trip_date'])
s710 = set(all_sched[all_sched['train_n'] == 710]['trip_date'])
cal['has_709'] = cal['trip_date'].isin(s709).astype(int)
cal['has_710'] = cal['trip_date'].isin(s710).astype(int)
```

```python
az_holidays = set(pd.to_datetime([
    '2024-01-01','2024-01-02','2024-01-03','2024-01-04','2024-01-05','2024-03-08',
    '2024-03-20','2024-03-21','2024-03-22','2024-03-23','2024-03-24','2024-03-25','2024-03-26',
    '2024-04-10','2024-04-11','2024-05-09','2024-05-28','2024-06-15','2024-06-16','2024-06-17',
    '2024-06-18','2024-06-19','2024-06-26','2024-11-08','2024-11-09','2024-12-31',
    '2025-01-01','2025-01-02','2025-01-03','2025-01-04','2025-01-05','2025-03-08',
    '2025-03-20','2025-03-21','2025-03-22','2025-03-23','2025-03-24','2025-03-25','2025-03-26',
    '2025-03-27','2025-03-28','2025-03-29','2025-03-30','2025-03-31','2025-05-09','2025-05-28',
    '2025-06-06','2025-06-07','2025-06-08','2025-06-09','2025-06-15','2025-06-16','2025-06-26',
    '2025-11-08','2025-11-09','2025-12-31',
    '2026-01-01','2026-01-02','2026-01-03','2026-01-04','2026-01-05','2026-03-08','2026-03-09',
    '2026-03-20','2026-03-21','2026-03-22','2026-03-23','2026-03-24','2026-03-25','2026-03-26',
    '2026-03-27','2026-03-28','2026-03-29','2026-03-30','2026-05-09','2026-05-10','2026-05-11',
    '2026-05-27','2026-05-28','2026-05-29','2026-05-30','2026-05-31','2026-06-15','2026-06-26'
]))
cal['is_holiday'] = cal['dt'].isin(az_holidays).astype(int)
cal['is_holiday_start'] = ((cal['is_holiday'] == 1) & (cal['is_holiday'].shift(1, fill_value=0) == 0)).astype(int)
cal['is_holiday_end'] = ((cal['is_holiday'] == 1) & (cal['is_holiday'].shift(-1, fill_value=0) == 0)).astype(int)
cal['is_pre_holiday'] = ((cal['is_holiday'] == 0) & (cal['is_holiday'].shift(-1, fill_value=0) == 1)).astype(int)
cal['is_post_holiday'] = ((cal['is_holiday'] == 0) & (cal['is_holiday'].shift(1, fill_value=0) == 1)).astype(int)
```

```python
# Contiguous block features for 709 & 710 runs (captures outbound vs inbound wave!)
for col, prefix in [('has_709', 'b709'), ('has_710', 'b710')]:
    sub = cal[cal[col] == 1][['trip_date', 'dt']].copy().reset_index(drop=True)
    gap = (sub['dt'] - sub['dt'].shift(1)).dt.days.fillna(999)
    bid = (gap > 1).cumsum()
    sub[f'{prefix}_len'] = sub.groupby(bid)['trip_date'].transform('count')
    sub[f'{prefix}_pos'] = sub.groupby(bid).cumcount()
    sub[f'{prefix}_dist_end'] = sub[f'{prefix}_len'] - 1 - sub[f'{prefix}_pos']
    sub[f'{prefix}_rel_pos'] = np.where(sub[f'{prefix}_len'] > 1, sub[f'{prefix}_pos'] / (sub[f'{prefix}_len'] - 1), 0.5)
    cal = cal.merge(sub.drop(columns=['dt']), on='trip_date', how='left')
    for f in [f'{prefix}_len', f'{prefix}_pos', f'{prefix}_dist_end']:
        cal[f] = cal[f].fillna(-1)
    cal[f'{prefix}_rel_pos'] = cal[f'{prefix}_rel_pos'].fillna(-1.0)
```

```python
def engineer(df):
    df = df.merge(cal, on='trip_date', how='left')
    df['is_weekend'] = (df['dow'] >= 5).astype(int)
    df['is_special_709_710'] = (
        ((df['train_n'] == 709) & (df['dow'] != 6)) |
        ((df['train_n'] == 710) & (df['dow'] != 5))
    ).astype(int)
    df['is_reduced_709_710_2025'] = (
        df['train_n'].isin([709, 710]) &
        (df['dt'] >= '2025-04-01') &
        (df['dt'] <= '2025-11-30')
    ).astype(int)
    df['seg_ratio'] = df['segment_order'] / df['n_segments']
    df['station_pair'] = df['from_station_id'].astype(str) + '_' + df['to_station_id'].astype(str)
    df['train_seg'] = df['train_n'].astype(str) + '_' + df['station_pair']
    return df
```

```python
# Filter to target trains + clean historical regime
train_df = train[train['train_n'].isin(target_trains)].copy()
train_df = engineer(train_df)
test_df = engineer(test.copy())

# Sample weights: metric weight (1 / n_segments) * recency weight * regime adjustment
train_df['w'] = 1.0 / train_df['n_segments']
# Upweight post-June 2025 data (matches 13-segment Qazax route & current demand regime)
train_df.loc[train_df['dt'] >= '2025-06-15', 'w'] *= 2.5
train_df.loc[train_df['dt'] >= '2025-12-01', 'w'] *= 1.5
# Downweight the anomalous 2025 Apr-Nov reduced-capacity period for 709/710
train_df.loc[train_df['is_reduced_709_710_2025'] == 1, 'w'] *= 0.15

cat_cols = ['train_n', 'route', 'direction', 'from_station_id', 'to_station_id', 'station_pair', 'train_seg']
num_cols = [
    'segment_order', 'n_segments', 'seg_ratio', 'capacity',
    'dow', 'month', 'is_weekend', 'has_709', 'has_710',
    'is_holiday', 'is_holiday_start', 'is_holiday_end', 'is_pre_holiday', 'is_post_holiday',
    'is_special_709_710', 'is_reduced_709_710_2025',
    'b709_len', 'b709_pos', 'b709_dist_end', 'b709_rel_pos',
    'b710_len', 'b710_pos', 'b710_dist_end', 'b710_rel_pos'
]
features = cat_cols + num_cols

for c in cat_cols:
    train_df[c] = train_df[c].astype('category')
    test_df[c] = test_df[c].astype('category')
```

```python
# GroupKFold by trip_date so all segments/trains on the same date stay in the same fold
gkf = GroupKFold(n_splits=5)
oof_lgb = np.zeros(len(train_df))
pred_lgb = np.zeros(len(test_df))

lgb_params = {
    'objective': 'regression',
    'metric': 'rmse',
    'learning_rate': 0.02,
    'num_leaves': 31,
    'max_depth': 6,
    'min_child_samples': 30,
    'subsample': 0.8,
    'subsample_freq': 1,
    'colsample_bytree': 0.8,
    'reg_alpha': 2.0,
    'reg_lambda': 5.0,
    'n_estimators': 3000,
    'random_state': 42,
    'n_jobs': -1,
    'verbose': -1
}
```

```python
for fold, (tr_idx, va_idx) in enumerate(gkf.split(train_df, train_df['occupancy'], groups=train_df['trip_date'])):
    X_tr, y_tr, w_tr = train_df.iloc[tr_idx][features], train_df.iloc[tr_idx]['occupancy'], train_df.iloc[tr_idx]['w']
    X_va, y_va, w_va = train_df.iloc[va_idx][features], train_df.iloc[va_idx]['occupancy'], train_df.iloc[va_idx]['w']

    m = lgb.LGBMRegressor(**lgb_params)
    m.fit(
        X_tr, y_tr, sample_weight=w_tr,
        eval_set=[(X_va, y_va)], eval_sample_weight=[w_va],
        callbacks=[lgb.early_stopping(100, verbose=False)]
    )
    oof_lgb[va_idx] = np.clip(m.predict(X_va), 0, None)
    pred_lgb += np.clip(m.predict(test_df[features]), 0, None) / 5.0

# Evaluate on recent post-2025-06-15 regime (closest to test distribution)
rec_mask = train_df['dt'] >= '2025-06-15'
w_metric = 1.0 / train_df.loc[rec_mask, 'n_segments']
print("Recent Regime OOF Weighted RMSE (LGBM):",
      round(root_mean_squared_error(train_df.loc[rec_mask, 'occupancy'], oof_lgb[rec_mask], sample_weight=w_metric), 5))

final_preds = pred_lgb.copy()
```

```python
if HAS_CATBOOST:
    oof_cat = np.zeros(len(train_df))
    pred_cat = np.zeros(len(test_df))
    train_cat = train_df.copy()
    test_cat = test_df.copy()
    for c in cat_cols:
        train_cat[c] = train_cat[c].astype(str)
        test_cat[c] = test_cat[c].astype(str)

    for fold, (tr_idx, va_idx) in enumerate(gkf.split(train_cat, train_cat['occupancy'], groups=train_cat['trip_date'])):
        cb = CatBoostRegressor(
            iterations=3500, learning_rate=0.03, depth=6, l2_leaf_reg=5.0,
            loss_function='RMSE', random_seed=42, verbose=0
        )
        cb.fit(
            train_cat.iloc[tr_idx][features], train_cat.iloc[tr_idx]['occupancy'],
            cat_features=cat_cols, sample_weight=train_cat.iloc[tr_idx]['w'],
            eval_set=(train_cat.iloc[va_idx][features], train_cat.iloc[va_idx]['occupancy']),
            early_stopping_rounds=150, verbose=False
        )
        oof_cat[va_idx] = np.clip(cb.predict(train_cat.iloc[va_idx][features]), 0, None)
        pred_cat += np.clip(cb.predict(test_cat[features]), 0, None) / 5.0

    oof_blend = 0.5 * oof_lgb + 0.5 * oof_cat
    print("Recent Regime OOF Weighted RMSE (Blend):",
          round(root_mean_squared_error(train_df.loc[rec_mask, 'occupancy'], oof_blend[rec_mask], sample_weight=w_metric), 5))
    final_preds = 0.5 * pred_lgb + 0.5 * pred_cat

# Post-processing: smooth identical-occupancy adjacent station blocks within each trip
test_df['occupancy'] = np.clip(final_preds, 0, None)
# For 731-734: Bakı-Dərnəgül-Biləcəri and Kürdəmir-Biləcəri share identical passenger counts
block_map = {
    '1_2': 'baki_bileceri', '2_3': 'baki_bileceri',
    '12_13': 'baki_bileceri', '11_12': 'baki_bileceri',
    '3_4': 'bileceri_ucar', '10_11': 'bileceri_ucar'
}
test_df['seg_key'] = test_df['segment_order'].astype(str) + '_' + (test_df['segment_order'] + 1).astype(str)
mask_73x = test_df['train_n'].astype(int).isin([731, 732, 733, 734])
test_df.loc[mask_73x, 'block_grp'] = test_df.loc[mask_73x, 'seg_key'].map(block_map)
valid_blk = mask_73x & test_df['block_grp'].notnull()
test_df.loc[valid_blk, 'occupancy'] = test_df.loc[valid_blk].groupby(['train_n', 'trip_date', 'block_grp'])['occupancy'].transform('mean')

sub = test_df[['id', 'occupancy']]
sub.to_csv('submission_v2.csv', index=False)
print("Saved submission_v2.csv successfully!")
```
